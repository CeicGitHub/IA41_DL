# Creacion de series

```python
import pandas as pd
import numpy as np
list1 = [83.9, 60, 70, 55]
serie_1 = pd.Series(data = list1,
                    index = ["Carlos", "Jimena", "Jesus", "Ana"],
                    name = "Peso",
                    dtype = np.float32,
                    copy = True) #
print(serie_1)
```

```python
serie_2 = pd.Series(data = list1)
print(serie_2)
```

```python
dic = {"alfa":12, "beta":15, "gamma":20, "delta":50}
print(dic)
series_3 = pd.Series(dic, dtype=np.float32)
print(series_3)
print(series_3.values)
```

```python
id = ["Carlos", "Jimena", "Jesus", "Ana"]
data = ["España", "Alemania", "Francia", "Holanda"]
series_4 = pd.Series(data, id, dtype = "string")
print(series_4)
```

```python
s5 = pd.Series([5, "?", np.nan, 8, pd.NaT, pd.NA, 6.9, True])
print(s5)
```

# Atributos

```python
dic = {"alfa":12, "beta":15, "gamma":20, "delta":50}
series_3 = pd.Series(dic, dtype=np.float32)
print(series_3)
print("Valores:",series_3.values)
print("Tipo de dato: ",series_3.dtype)
print("Indices: ",series_3.index)
print("Forma:", series_3.shape)
print("Size: ",series_3.size)
print("Nombre: ",series_3.name)
```

# Indexacion 

Indexacion posicional

```python
# indexacion posicional
print(series_3)
print("-------indexacion simple")
print(series_3.iloc[1])   # Recomendado
print(series_3.iat[1])    # Mas recomendado
print(series_3[1])        # No recomendado
print("-----idexacion multiple")
print(series_3.iloc[[0, 3]])    # Recomendado
#print(series_3.iat[[0, 3]])    # Error - iat no funciona con indexacion multple
print(series_3[[0, 3]])         # No recomendado
print("-----idexaxion slicing")
print(series_3.iloc[0:2])       # Recomendado
#print(series_3.iat[0:2])       # Error - iat no funciona con indexacion slicing
print(series_3[0:2])            # No recomendado
```

Indexacion por etiqueda

```python
dic = {"beta":15, "gamma":20, "delta":50, "alfa":12}
series_5 = pd.Series(dic, dtype=np.float32)
print(series_5)
print("-----Indexacion simple")
print(series_5.loc["alfa"])
print(series_5.at["alfa"])
print("-----Indexacion multiple")
print(series_5.loc[["alfa", "delta"]])
print("-----Indexacion slicing")
print(series_5.loc["beta": "gamma"])
```

indexacion condicional

```python
print(series_5)
print("---------------")
print(series_5[series_5 > 15])                     #convencional
print("---------------")
print(series_5.loc[series_5 > 15])
```

# Metodos de modificacion

reemplazar elementos

```python
dic = {"beta":15, "gamma":20, "delta":50, "alfa":12}
S5 = pd.Series(dic, dtype=np.float32)
```

Reemplazar un elemento

```python
print(S5)
print("Reemplazar por posicion")
S5.iloc[0] = 150
print("Reemplazar por etiqueta")
S5.loc["gamma"] = 200
print("Reemplazar por valor")
S5.replace(50,500, inplace = True)
print(S5)
```

Reemplazar varios elementos

```python
dic = {"beta":15, "gamma":20, "delta":50, "alfa":12}
S5 = pd.Series(dic, dtype=np.float32)
print(S5)
print("Reemplazar por posicion")
S5.iloc[[0,3]] = [150, 120]
print("Reemplazar por etiqueta")
S5.loc[["gamma","delta"]] = [200, 500]
print(S5)
print("Reemplazar por valor")
S5.replace([150,500],[1.5,5], inplace = True)
print(S5)
```

Replace vs asignacion

```python
id = ["Azul", "Verde", "Cafe"]
S9 = pd.Series([1, 2, 3],id)
print("Vista de S9")
print(S9.replace(1,100))
print("S9 real")
print(S9)
```

```python
id = ["Azul", "Verde", "Cafe"]
S9 = pd.Series([1, 2, 3],id)
S11 = S9.replace(1,100)                 #asignacion
print("Guarde la lista de S9 en S11")
print(S11)
print("S9 real")
print(S9)
```

```python
id = ["Azul", "Verde", "Cafe"]
S9 = pd.Series([1, 2, 3],id)
S9.replace(1,100, inplace = True)                 #ocupando replace
print(S9)
```

copia vs asignacion

```python
arr = np.array([1, 2, 3, 4, 5, 6])
S11 = pd.Series(arr)
arr[0] = 100
print("Arreglo: ", arr)
print("Serie:\n ",S11)
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])
S11 = pd.Series(arr, copy = True)
arr[0] = 100
print("Arreglo: ", arr)
print("Serie:\n ",S11)
```

```python
arr = np.array([1, 2, 3])
S11 = pd.Series(arr, copy = True)
S12 = S11
S12.iloc[2] = 300
print(S12)
print(S11)
```

```python
arr = np.array([1, 2, 3])
S11 = pd.Series(arr, copy = True)
S12 = S11.copy()
S12.iloc[2] = 300
print(S12)
print(S11)
```

agregar elementos

```python
#Agregar un elemento
id = ["Azul", "Verde", "Cafe"]
S9 = pd.Series([1, 2, 3],id)
S9.loc["Amarillo"] = 10
print(S9)
```

```python
#arregar varios elementos
id = ["Azul", "Verde", "Cafe"]
S9 = pd.Series([1, 2, 3],id)
S10 = pd.Series([4, 5], ["Rosa", "Morado"])
S11 = pd.concat([S9, S10])
print(S11)
```

eliminar objetos

```python
id = ["Azul", "Verde", "Cafe","Rosa", "Morado"]
S9 = pd.Series([1, 2, 3, 4, 5],id)
S9.drop("Verde", inplace = True)
S9.drop(["Azul","Morado"], inplace = True)
print(S9)
print("---", S9.pop("Rosa"))
print(S9)
```

cambiar indices

```python
id = ["Azul", "Verde", "Cafe"]
S9 = pd.Series([1, 2, 3],id)
print("Original")
print(S9)
S9.index = ["x", "y", "z"]
print("Etiquetas modificadas con el metodo index")
print(S9)
S9 = S9.set_axis(["i", "j", "k"])
print("Etiquetas modificadas con el metodo set_axis")
print(S9)
S9 = S9.rename({"i":"I","k":"K"})
print("Etiquetas modificadas con el metodo rename")
print(S9)
S9 = S9.reindex(["I", "j", "K", "z"])
print("Etiquetas modificadas con el metodo reindex")
print(S9)
```

ordenar

```python
#por valor
import pandas as pd
s1 = pd.Series([80, 40, 60],["a", "b", "c"])
s2 = pd.Series(["France","Germany","Brazil"],["x","y","z"])
print(s1)
print(s2)
print("orden ascendente")
s1.sort_values(inplace=True)
s2.sort_values(inplace=True)
print("S1")
print(s1)
print("S2")
print(s2)
print("orden descendente")

s1.sort_values(inplace=True,ascending=False)
s2.sort_values(inplace=True,ascending=False)
print("S1")
print(s1)
print("S2")
print(s2)
s3=s1.sort_values(ignore_index=True)
print("S3")
print(s3)
```

```python
#por etiqueta
import pandas as pd
s1 = pd.Series([1, 2, 3],["c", "a", "b"])
s2 = pd.Series(["France","Germany","Brazil"],[1,3,2])
print(s1)
print(s2)
print("orden ascendente")
s1.sort_index(inplace=True)
s2.sort_index(inplace=True)
print("S1")
print(s1)
print("S2")
print(s2)
print("orden descendente")
s1.sort_index(inplace=True,ascending=False)
s2.sort_index(inplace=True,ascending=False)
print("S1")
print(s1)
print("S2")
print(s2)
s3=s1.sort_index(ignore_index=True)
print(s3)
```

Duplicados

```python
pd.set_option('future.no_silent_downcasting', True)
import pandas as pd
import numpy as np
data=[0.8,2,np.nan,3.2,None,4.8,"x"]
id=["a","b","c","d","e","a","f"]
s1=pd.Series(data,id)
s1=s1.dropna()              #elimino datos nulos
s1=s1.replace("x",0)        #reemplazo caracteres raros por numeros
s1=s1.astype("float64")     #convierto de serie de tipo object a flotante64
print("S1 Limpia:")
print(s1)
#buscamos cuantos duplicados(indices) tenemos 
print(s1.index.duplicated(keep="last"))
print(s1.index.duplicated())
print(~s1.index.duplicated())
#identificamos el indice duplicado
print(s1.index[s1.index.duplicated()])
print(s1[s1.index.duplicated()])
print(s1[~s1.index.duplicated()])

#Se obetiene el promedio de los datos duplicados
print(s1.groupby(level=0).mean()) #or sum(),min(),max()
```
